# Steps to Create a Jupyter Notebook in the llm Folder
This notebook demonstrates how to set up the `llm` folder, create a Jupyter notebook file, verify its existence, and open/edit it in Visual Studio Code.

## 1. Set Up the llm Folder Structure
Use Python's `os` module to check for and create the `llm` folder if it does not exist.

````python
import os
folder_path = 'llm'
if not os.path.exists(folder_path):
    os.makedirs(folder_path)
print(f"Folder '{folder_path}' is ready.")
````

## 2. Create a Jupyter Notebook File (.ipynb)
Use Python to programmatically create an empty `.ipynb` file named `jpunb.ipynb` inside the `llm` folder.

````python
import nbformat as nbf
notebook_path = os.path.join(folder_path, 'jpunb.ipynb')
if not os.path.exists(notebook_path):
    nb = nbf.v4.new_notebook()
    nbf.write(nb, notebook_path)
print(f"Notebook '{notebook_path}' created.")
````

## 3. Verify Notebook Creation in llm Folder
List the contents of the `llm` folder to confirm that `jpunb.ipynb` exists.

````python
print("Contents of 'llm' folder:", os.listdir(folder_path))
````

## 4. Open and Edit the Notebook in Visual Studio Code
You can open and start editing `llm/jpunb.ipynb` directly in Visual Studio Code. Use the file explorer or run the following command in the terminal:

```bash
code llm/jpunb.ipynb
```

This will open the notebook for editing.
